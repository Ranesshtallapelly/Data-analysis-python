```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os # Import the os module

# Create the directory if it doesn't exist
output_dir = "/mnt/data"
os.makedirs(output_dir, exist_ok=True)

# -------- Q1: Simulate Poisson λ=10 for 30 days --------
lam = 10
days = 30
np.random.seed(42)
data_q1 = np.random.poisson(lam, days)

df_q1 = pd.DataFrame({
    "day": np.arange(1, days+1),
    "emails_received": data_q1
})

q1_path = os.path.join(output_dir, "poisson_lambda10_30days.csv")
df_q1.to_csv(q1_path, index=False)

# Plot histogram
plt.figure(figsize=(8,5))
plt.hist(data_q1, bins=10)
plt.title("Poisson Simulation: λ = 10 for 30 Days")
plt.xlabel("Emails per day")
plt.ylabel("Frequency")
plot1_path = os.path.join(output_dir, "poisson_hist_lambda10.png")
plt.savefig(plot1_path)
plt.show()


# -------- Q2: λ = [2,5,10,15,25] with 1000 samples each --------
lam_values = [2,5,10,15,25]
samples = {}

plt.figure(figsize=(15,10))
for i, lam in enumerate(lam_values, 1):
    sim = np.random.poisson(lam, 1000)
    samples[lam] = sim

    plt.subplot(3,2,i)
    plt.hist(sim, bins=20)
    plt.title(f"λ = {lam}")
    plt.xlabel("Emails")
    plt.ylabel("Frequency")

plot2_path = os.path.join(output_dir, "poisson_lam_all_subplots.png")
plt.tight_layout()
plt.savefig(plot2_path)
plt.show()

# Save Q2 datasets
for lam, sim in samples.items():
    pd.DataFrame({"emails": sim}).to_csv(os.path.join(output_dir, f"poisson_lambda{lam}.csv"), index=False)

print("Files generated!")
```
