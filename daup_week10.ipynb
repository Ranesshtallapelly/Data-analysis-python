```python
import math, numpy as np, pandas as pd, matplotlib.pyplot as plt, json
import os # Import the os module
np.random.seed(42)

n = 15
p = 0.3

def binomial_pmf(n,k,p):
    return math.comb(n,k) * (p**k) * ((1-p)**(n-k))

rows = []
cum = 0.0
for k in range(0, n+1):
    pmf = binomial_pmf(n,k,p)
    cum += pmf
    rows.append({"k": k, "pmf": pmf, "cdf": cum, "n": n, "p": p, "type": "pmf_cdf"})

pmf_df = pd.DataFrame(rows)

samples = np.random.binomial(n, p, size=20)
samples_df = pd.DataFrame({
    "k": np.arange(1, 21),
    "opens": samples,
    "pmf": samples,
    "cdf": [None]*20,
    "n": [n]*20,
    "p": [p]*20,
    "type": ["samples"]*20
})

pmf_df = pmf_df[["k","pmf","cdf","n","p","type"]]
samples_df = samples_df[["k","pmf","cdf","n","p","type"]]
combined = pd.concat([pmf_df, samples_df], ignore_index=True)

# Create the directory if it doesn't exist
os.makedirs('/mnt/data', exist_ok=True)

combined.to_csv("/mnt/data/week10_binomial_full_combined.csv", index=False)
pmf_df.to_csv("/mnt/data/week10_binomial_pmf.csv", index=False)
samples_df.to_csv("/mnt/data/week10_binomial_20_samples.csv", index=False)

plt.figure(figsize=(8,5))
plt.stem(pmf_df["k"], pmf_df["pmf"])
plt.title("Binomial PMF (n=15, p=0.3)")
plt.xlabel("k (number of opens)")
plt.ylabel("P(X=k)")
plt.grid(True)
plt.tight_layout()
plt.savefig("/mnt/data/week10_pmf_plot.png")
plt.close()

plt.figure(figsize=(8,5))
plt.step(pmf_df["k"], pmf_df["cdf"], where='mid')
plt.title("Binomial CDF (n=15, p=0.3)")
plt.xlabel("k (number of opens)")
plt.ylabel("P(X ≤ k)")
plt.grid(True)
plt.tight_layout()
plt.savefig("/mnt/data/week10_cdf_plot.png")
plt.close()

print("Saved datasets and plots to /mnt/data")
```
